# **Performance Comparison: Python Lists vs NumPy Arrays**

This notebook benchmarks the execution time of squaring one million numbers using:

1. A standard Python list with a list comprehension  
2. A NumPy array using vectorized operations  

### Goal
Demonstrate why NumPy is significantly faster than pure Python when working with large numerical datasets.

### Concepts Practiced
- Creating arrays with `np.array`
- Measuring execution time using `time.time()`
- Vectorized mathematical operations in NumPy
- Understanding why NumPy is optimized for numerical computing

This small experiment highlights one of the key advantages of NumPy: performance.

```python
import numpy as np
import time
```

```python
test_list = [1, 2, 3, 4, 5]
test_array = np.array(test_list)
print('List: ', test_list)
print('Array: ', test_array)
```

```python
# Creates a list with 1,000,000 elements
data_list = list(range(1000000))

# Converts the list into a NumPy array
data_array = np.array(data_list)

# Starts timing the operation on the list
start_time = time.time()

# Performs the operation of squaring each element in the list
list_squared = [i**2 for i in data_list]

# Stops the timer
list_time = time.time() - start_time

# Starts timing the operation on the array
start_time = time.time()

# Performs the operation of squaring each element in the array
array_squared = data_array**2

# Stops the timer
array_time = time.time() - start_time

print("Time for the list operation: ", list_time)
print("Time for the array operation: ", array_time)
```
